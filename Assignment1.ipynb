```python
import cv2 as cv
import matplotlib.pyplot as plt
```

```python
#Read an image
image=cv.imread("image.jpg")
```

```python
#Display an image
plt.imshow(image)
```

```python
#Write an image
cv.imwrite("output.jpg",image)
```

```python
#Converting image into different formats
cv.imwrite("output.png",image)
cv.imwrite("output.tiff",image)
cv.imwrite("output.bmp",image)
```

```python
#Resizing an image
resize= cv.resize(image,(400,300))
plt.imshow(resize)
```

```python
#Convert to grayscale
gray= cv.cvtColor(image,cv.COLOR_BGR2GRAY)
plt.imshow(gray)
```
