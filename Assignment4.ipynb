```python
#1. WAP to enhance the input image using Log transform. Assume c=1.
#FORMULA FOR LOG TRANSFORMATION - s=c×log(1+r)
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("image.jpg", 0)

img = img.astype(float)

log_img = np.log(1 + img)

log_img = np.uint8(255 * log_img / np.max(log_img))

plt.imshow(log_img, cmap="gray")
plt.axis("off")
plt.show()
```

```python
#2. WAP to enhance the input image using Power law/gamma transform. Assume c=1, and input the value of gamma from the user.  
#FORMULA = s=c×rγ
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("image.jpg", 0)

gamma = float(input("Enter gamma value: "))

gamma_img = 255 * (img / 255) ** gamma

gamma_img = np.uint8(gamma_img)

plt.imshow(gamma_img, cmap="gray")
plt.axis("off")
plt.show()
```

```python
#3. WAP to enhance the input image using Intensity level slicing (Contrast stretching). The input range of intensity values and the desired range of intensity values in the output image are obtained from the user. (Example: The input range of intensity values from 80 to 120 is stretched to 50-150 in the output image) 
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("image.jpg", 0)

a = int(input("Enter input lower value: "))
b = int(input("Enter input upper value: "))

c = int(input("Enter output lower value: "))
d = int(input("Enter output upper value: "))

output = img.copy()

# Stretch only the selected range
mask = (img >= a) & (img <= b)

output[mask] = ((img[mask] - a) * (d - c) / (b - a) + c)

output = output.astype(np.uint8)

plt.imshow(output, cmap="gray")
plt.axis("off")
plt.show()
```
