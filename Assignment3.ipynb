```python
#1.WAP TO DRAW A BORDER AROUND THE INPUT IMAGE AND INPUT THE DESIRED COLOR OF BORDER FROM THE USER.
#(FOR GRAYSCALE IMAGE)
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("image.jpg", 0)

width = int(input("Enter border width: "))

r = int(input("Enter R value: "))
g = int(input("Enter G value: "))
b = int(input("Enter B value: "))

# Convert grayscale to color
img = cv2.cvtColor(img, cv2.COLOR_GRAY2BGR)

# Add border
border = cv2.copyMakeBorder(img, width, width, width, width,
                            cv2.BORDER_CONSTANT, value=(b, g, r))

# Convert BGR to RGB
border = cv2.cvtColor(border, cv2.COLOR_BGR2RGB)

plt.imshow(border)
plt.axis("off")
plt.show()
```

```python
#(FOR BINARY IMAGE)
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("image.jpg", 0)

# Convert to binary
_, binary = cv2.threshold(img, 128, 255, cv2.THRESH_BINARY)

# Take inputs
width = int(input("Enter border width: "))

r = int(input("Enter R value: "))
g = int(input("Enter G value: "))
b = int(input("Enter B value: "))

# Convert binary to color
binary = cv2.cvtColor(binary, cv2.COLOR_GRAY2BGR)

# Add border
border = cv2.copyMakeBorder(binary, width, width, width, width,
                            cv2.BORDER_CONSTANT, value=(b, g, r))

# Convert BGR to RGB for matplotlib
border = cv2.cvtColor(border, cv2.COLOR_BGR2RGB)

plt.imshow(border)
plt.axis("off")
plt.show()
```

```python
#2. WAP FOR IMAGE COMPLIMENT
#(FOR GRAYSCALE IMAGE)
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("image.jpg", 0)

complement = 255 - img

plt.imshow(complement, cmap="gray")
plt.axis("off")
plt.show()
```

```python
#(FOR BINARY IMAGE)
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("image.jpg", 0)

# Convert to binary
_, binary = cv2.threshold(img, 128, 255, cv2.THRESH_BINARY)

# Complement
complement = 255 - binary

plt.imshow(complement, cmap="gray")
plt.axis("off")
plt.show()
```
